```python
from matplotlib import pyplot as plt
%matplotlib widget
import os
import tyro
from pathlib import Path
import ipynbname
os.chdir(ipynbname.path().parent)

import fit_latents

base_dir = Path('<path to where data and models are stored (see README)')
model_path = base_dir / 'models/unconditioned/checkpoints/checkpoint_5000.pt'
autoencoder_path = base_dir / 'models/autoencoder/checkpoints/checkpoint_4500.pt'
dataset_path = base_dir / 'dataset/sparse_scans_20251022_120753/datalist.json'
config_path = 'configs/sparse/config_ggl.yaml'

with open(config_path, 'r') as yaml_file:
	config = tyro.extras.from_yaml(fit_latents.TrainingConfig, yaml_file)

overrides = dict(
	output_dir=Path('results'),
	experiment_name='experiment',
	pointcloud_path=dataset_path,
	model_path=model_path,
	autoencoder_path=autoencoder_path,
	save_animation=False,
	gt_scale=False,
	save_mesh=True,
	mc_resolution=256,
)

for key, value in overrides.items():
	setattr(config, key, value)

experiment_dir, results = fit_latents.run(config)
```

## Parse results

```python
import pandas as pd
import json
experiment_dir = Path('/dss/dssfs02/lwp-dss-0001/pr28vo/pr28vo-dss-0001/ge83xit2/diffusion_experiments/results_gg/experiment_2026-03-20_16:13:40/')

with open(experiment_dir / 'datalist.json') as json_file:
	datalist = json.load(json_file)

results = []
for datapoint in datalist:
	metrics_path = experiment_dir / datapoint['model_id'] / 'eval_metrics.csv'
	results.append(pd.read_csv(experiment_dir / datapoint['model_id'] / 'eval_metrics.csv', index_col=0).assign(
		model_id=datapoint['model_id'],
	))
results = pd.concat(results, ignore_index=True)
print(f'CD: {results.chamfer_distance.mean() * 100}')
print(f'CA: {results.chamfer_unoriented_normal_angle.mean()}')
results
```
